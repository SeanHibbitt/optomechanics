```python
def OMIT_equation(f, f_m, g_m, A, floor, f_c, kk_tot):

    """
    OMIT fitting equation, A is defined as:
    A = (k_ext * 4 * omega_m * g_0**2 * n_d)  /  (sqrt(k**4 + 4*(w_p - w_c)) ))
    floor = sqrt(k_ext**2 / (sqrt(k**4 + 4*(w_p - w_c))  )

    """

    w_p = 2*np.pi* f
    w_m = 2*np.pi*f_m
    w_c = 2*np.pi*f_c
    G_m = 2*np.pi*g_m
    k_tot = 2*np.pi*kk_tot

    denominator = (w_m**2 - (w_p - w_c)**2 - 1j*G_m*(w_p - w_c)) * (k_tot - 2j*(w_p-w_c))

    OMIT =  floor*(1+ A * np.abs( 1/ denominator) )
    
    return OMIT
```

```python
f_c = 0
k_tot = 10e6

freqs = np.linspace(95e3, 105e3, int(10e3))

f_m = 100e3
g_m = 100
A = -dB_to_linear_voltage(327)
floor = dB_to_linear_voltage(-18.2)

plt.plot(freqs, linear_voltage_to_dB(OMIT_equation(freqs, f_m, g_m, A, floor, f_c, k_tot )))
```
