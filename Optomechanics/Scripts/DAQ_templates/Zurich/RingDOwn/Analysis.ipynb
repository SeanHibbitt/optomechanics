# transform and plot data

```python
# try:
#     plt.close(fig)
# except: pass

fig, ax = plt.subplots()
ax.set(title = "Ring-down Measurements for multiple time constants", xlabel = "Time[s]", ylabel = "Normalised Amplitude[]")

fig_plotly_1 = go.Figure()
fig_plotly_1.update_layout(
title="Ring-down Measurements for multiple time constants",
xaxis_title="Time[s]",
yaxis_title="Normalised Amplitude[]"
)

def Q(tau, f_m):

    return tau*f_m*np.pi


for resonance_loc in data_dict.keys():
    for tc in data_dict[resonance_loc].keys():
        time, amplitude = data_dict[resonance_loc][tc] 

        
        amplitude = amplitude/np.max(amplitude)

        time = time[:-1:100]
        amplitude = amplitude[:-1:100]

        amplitude_avg_rolled = []
        rolling_avg_num = 1000
        
        for i in range(amplitude.size):
            try:
                amplitude_avg_rolled.append(np.mean(amplitude[i:i+rolling_avg_num]))
            except:
                amplitude_avg_rolled.append(amplitude[i])

        reduced_indicies = time >= time[np.argmax(amplitude_avg_rolled)]

        time = time[reduced_indicies]
        amplitude = amplitude[reduced_indicies]
        amplitude_avg_rolled = np.array(amplitude_avg_rolled)[reduced_indicies]/np.max(amplitude_avg_rolled)

        reduced_indicies_2 = amplitude_avg_rolled <= 0.4
        time = time[reduced_indicies_2]
        amplitude = amplitude[reduced_indicies_2]
        amplitude_avg_rolled = np.array(amplitude_avg_rolled)[reduced_indicies_2]
        
        time = time-time[0]

        reduced_indicies_3 = time <= 24
        time = time[reduced_indicies_3]
        amplitude = amplitude[reduced_indicies_3]
        amplitude_avg_rolled = np.array(amplitude_avg_rolled)[reduced_indicies_3]


        popt, pcov, guesses = fit_ring_down(time, amplitude)


        #f_m taken from photos
        if "6" in resonance_loc:
            f_m = 6.1287605e3
        if "14" in resonance_loc:
            f_m = 14.4612011e3
        Q = tau_to_Q(popt[1], f_m)

        amplitude_fit = ring_down_eq(time, *popt)

        ax.plot(time, amplitude/popt[0], label = f"{resonance_loc}_{tc}ms")
        ax.plot(time, amplitude_fit/popt[0], label = f"fit:{resonance_loc}_{tc}ms, Q = {Q}, popt:{popt}")

        fig_plotly_1.add_trace(go.Scatter(x=time, y=amplitude/popt[0], mode='lines', name=f"{resonance_loc}_{tc}ms"))
        fig_plotly_1.add_trace(go.Scatter(x=time, y=amplitude_fit/popt[0], mode='lines', name=f"fit:{resonance_loc}_{tc}ms, Q = {Q:.3e}, popt:{popt}"))



ax.legend()
fig.show()

fig_plotly_1.write_html(f"ring_down.html")
```
